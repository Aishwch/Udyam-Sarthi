```python
import pandas as pd
import numpy as np

print("Dataset environment ready ✅")
```

```python
# Create machine information

machines = pd.DataFrame({
    "machine_id": ["M001", "M002", "M003", "M004", "M005"],
    
    "type": [
        "CNC",
        "CNC",
        "Press",
        "Lathe",
        "Milling"
    ],
    
    "age": [
        2,
        5,
        3,
        7,
        1
    ],
    
    "rated_capacity": [
        100,
        120,
        150,
        90,
        110
    ],
    
    "power_rating": [
        50,
        60,
        70,
        45,
        55
    ],
    
    "installation_date": [
        "2024-01-15",
        "2021-06-10",
        "2023-03-20",
        "2019-09-05",
        "2025-02-12"
    ]
})

machines
```

```python
# Check our machine data

print("Number of machines:", len(machines))
print("\nMachine information:")
display(machines)
```

```python
# Generate synthetic daily operations data

np.random.seed(42)

dates = pd.date_range(
    start="2026-01-01",
    end="2026-06-30",
    freq="D"
)

shifts = ["Shift-1", "Shift-2", "Shift-3"]

operations = []

for date in dates:
    for _, machine in machines.iterrows():
        for shift in shifts:

            # Random operating conditions
            runtime = np.random.uniform(5.5, 8.0)
            downtime = 8.0 - runtime

            # Expected production is based on runtime and machine capacity
            expected_output = runtime * machine["rated_capacity"]

            # Small production variation
            production_factor = np.random.normal(0.95, 0.05)

            actual_output = expected_output * production_factor

            # Energy is related to runtime and machine power
            energy_kwh = runtime * machine["power_rating"]
            energy_kwh *= np.random.normal(1.0, 0.05)

            # Scrap is a small percentage of production
            scrap_units = actual_output * np.random.uniform(0.01, 0.04)

            operations.append({
                "date": date,
                "machine_id": machine["machine_id"],
                "shift": shift,
                "runtime": round(runtime, 2),
                "downtime": round(downtime, 2),
                "expected_output": round(expected_output, 2),
                "actual_output": round(actual_output, 2),
                "energy_kwh": round(energy_kwh, 2),
                "scrap_units": round(scrap_units, 2)
            })

daily_operations = pd.DataFrame(operations)

print("Daily operations created successfully! ✅")
print("Number of records:", len(daily_operations))

display(daily_operations.head(10))
```

```python
# Add machine capacity and power information

daily_check = daily_operations.merge(
    machines[["machine_id", "rated_capacity", "power_rating"]],
    on="machine_id",
    how="left"
)

display(daily_check.head())
```

```python
# Calculate expected production from runtime and machine capacity

daily_check["calculated_expected"] = (
    daily_check["runtime"] *
    daily_check["rated_capacity"]
)

display(
    daily_check[
        [
            "machine_id",
            "runtime",
            "rated_capacity",
            "expected_output",
            "calculated_expected"
        ]
    ].head(10)
)
```

```python
# Calculate expected energy from runtime and machine power

daily_check["calculated_energy"] = (
    daily_check["runtime"] *
    daily_check["power_rating"]
)

display(
    daily_check[
        [
            "machine_id",
            "runtime",
            "power_rating",
            "energy_kwh",
            "calculated_energy"
        ]
    ].head(10)
)
```

```python
# Create a separate dataset for ML features

ml_data = daily_check.copy()

print("ML feature dataset created ✅")
print("Rows:", len(ml_data))
```

```python
# Production variance

ml_data["production_variance"] = (
    ml_data["actual_output"] -
    ml_data["expected_output"]
)

display(
    ml_data[
        [
            "expected_output",
            "actual_output",
            "production_variance"
        ]
    ].head(10)
)
```

```python
# Production variance percentage

ml_data["production_variance_pct"] = (
    ml_data["production_variance"] /
    ml_data["expected_output"]
) * 100

display(
    ml_data[
        [
            "expected_output",
            "actual_output",
            "production_variance_pct"
        ]
    ].head(10)
)
```

```python
# Machine utilization

shift_duration = 8

ml_data["utilization_pct"] = (
    ml_data["runtime"] /
    shift_duration
) * 100

display(
    ml_data[
        [
            "runtime",
            "utilization_pct"
        ]
    ].head(10)
)
```

```python
# Downtime rate

ml_data["downtime_rate_pct"] = (
    ml_data["downtime"] /
    shift_duration
) * 100

display(
    ml_data[
        [
            "downtime",
            "downtime_rate_pct"
        ]
    ].head(10)
)
```

```python
# Scrap rate

ml_data["scrap_rate_pct"] = (
    ml_data["scrap_units"] /
    ml_data["actual_output"]
) * 100

display(
    ml_data[
        [
            "actual_output",
            "scrap_units",
            "scrap_rate_pct"
        ]
    ].head(10)
)
```

```python
# Energy intensity

ml_data["energy_intensity"] = (
    ml_data["energy_kwh"] /
    ml_data["actual_output"]
)

display(
    ml_data[
        [
            "energy_kwh",
            "actual_output",
            "energy_intensity"
        ]
    ].head(10)
)
```

```python
# Check for impossible negative values

print("Negative runtime:", (ml_data["runtime"] < 0).sum())
print("Negative downtime:", (ml_data["downtime"] < 0).sum())
print("Negative expected output:", (ml_data["expected_output"] < 0).sum())
print("Negative actual output:", (ml_data["actual_output"] < 0).sum())
print("Negative energy:", (ml_data["energy_kwh"] < 0).sum())
print("Negative scrap:", (ml_data["scrap_units"] < 0).sum())
```

```python
# Check for duplicate records

duplicates = ml_data.duplicated().sum()

print("Duplicate records:", duplicates)
```

```python
# Check date data type

print(ml_data["date"].dtype)
```

```python
# Sort data chronologically

ml_data = ml_data.sort_values("date").reset_index(drop=True)

print("Data sorted by date ✅")
print("First date:", ml_data["date"].min())
print("Last date:", ml_data["date"].max())
```

```python
# Select features (X) and target (y)

features = [
    "runtime",
    "downtime",
    "expected_output",
    "energy_kwh",
    "scrap_units"
]

X = ml_data[features]
y = ml_data["actual_output"]

print("Features (X):")
print(features)

print("\nTarget (y):")
print("actual_output")

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Check our date range

print("First date:", ml_data["date"].min())
print("Last date:", ml_data["date"].max())
print("Total records:", len(ml_data))
```

```python
# Time-based train-test split

split_index = int(len(ml_data) * 0.8)

train_data = ml_data.iloc[:split_index]
test_data = ml_data.iloc[split_index:]

print("Training records:", len(train_data))
print("Testing records:", len(test_data))

print("\nTraining period:")
print(train_data["date"].min(), "to", train_data["date"].max())

print("\nTesting period:")
print(test_data["date"].min(), "to", test_data["date"].max())
```

```python
# Create a clean date-based train-test split

unique_dates = sorted(ml_data["date"].unique())

split_date_index = int(len(unique_dates) * 0.8)
split_date = unique_dates[split_date_index]

train_data = ml_data[ml_data["date"] < split_date]
test_data = ml_data[ml_data["date"] >= split_date]

print("Training records:", len(train_data))
print("Testing records:", len(test_data))

print("\nTraining period:")
print(train_data["date"].min(), "to", train_data["date"].max())

print("\nTesting period:")
print(test_data["date"].min(), "to", test_data["date"].max())
```

```python
# Calculate the baseline prediction

baseline_prediction = train_data["actual_output"].mean()

print("Baseline prediction:", round(baseline_prediction, 2))
```

```python
# Create baseline predictions for the test data

baseline_predictions = np.full(
    len(test_data),
    baseline_prediction
)

print("Number of baseline predictions:", len(baseline_predictions))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

baseline_mae = mean_absolute_error(
    test_data["actual_output"],
    baseline_predictions
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        test_data["actual_output"],
        baseline_predictions
    )
)

print("Baseline MAE:", round(baseline_mae, 2))
print("Baseline RMSE:", round(baseline_rmse, 2))
```

```python
# Prepare training and testing features

X_train = train_data[features]
y_train = train_data["actual_output"]

X_test = test_data[features]
y_test = test_data["actual_output"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Create the model
linear_model = LinearRegression()

# Teach the model using training data
linear_model.fit(X_train, y_train)

print("Linear Regression trained successfully! ✅")
```

```python
# Make predictions on unseen test data

linear_predictions = linear_model.predict(X_test)

print("Predictions created:", len(linear_predictions))
print("\nFirst 10 predictions:")
print(linear_predictions[:10])
```

```python
# Evaluate Linear Regression

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

print("Linear Regression MAE:", round(linear_mae, 2))
print("Linear Regression RMSE:", round(linear_rmse, 2))
```

```python
from sklearn.ensemble import RandomForestRegressor

print("Random Forest library ready ✅")
```

```python
# Create Random Forest model

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Random Forest model created ✅")
```

```python
# Train Random Forest

rf_model.fit(X_train, y_train)

print("Random Forest trained successfully! ✅")
```

```python
# Make predictions on test data

rf_predictions = rf_model.predict(X_test)

print("Random Forest predictions created:", len(rf_predictions))
```

```python
# Evaluate Random Forest

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

print("Random Forest MAE:", round(rf_mae, 2))
print("Random Forest RMSE:", round(rf_rmse, 2))
```
